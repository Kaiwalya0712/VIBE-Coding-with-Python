### Object Oriented Programming

```python
class Person:
    def __init__(self, fname):
        self.name = fname

    def say_hi(self):
        print('Hello, my name is', self.name)
```

```python
p=Person('John')
```

```python
q=Person('Jane')
```

```python
p.say_hi()
q.say_hi()
```

### Classes: Grouping data


```python
class Time:
    def __init__(self):
        self.hours = 0
        self.minutes = 0

    def print_time(self):
        print('Hours:', self.hours, end='')
        print('Minutes:', self.minutes)
```
